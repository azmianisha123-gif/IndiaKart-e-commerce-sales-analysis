## Load all csv file

```python
import zipfile
import pandas as pd
import numpy as np

#Path to the zip file
zip_path = r"c:\Users\hp\Downloads\44120013dcc669c383d89c8c4a40b746.zip"

#Dictionary to store all csv files
dataframes = {}
with zipfile.ZipFile(zip_path, 'r') as z:
    #List all the files in the zip
    display(z.namelist())

    #Read only csv files
    for file in z.namelist():
        if file.endswith('.csv'):
            with z.open(file) as f:
                df = pd.read_csv(f)
                dataframes[file] = df
#Display the first five rows of each csv                
for name, df in dataframes.items():
    display(f"\n {name}:")
    display(df.head())
```

# Basic Data Profiling :

```python
# 2. For each table: row count, column count, data types
for name, df in dataframes.items():
    display(f"\nProfile for {name}:")
    display(f"Row count: {df.shape[0]}")
    display(f"Column count: {df.shape[1]}")
    display(f"Data types:\n{df.dtypes}")
```

# Missing value Analysis

```python
## 3. Missing values in every column —columns have nulls and % of rows affected.
print("="*70)
print("MISSING VALUE ANALYSIS")
print("="*70)

for name, df in dataframes.items():

    print(f"\n{name}")

    missing = df.isna().sum()

    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values.")
        continue

    result = pd.DataFrame({
        "Missing Values": missing,
        "% Rows Affected": (missing / len(df) * 100).round(2)
    })

    print(result)
```

# Duplicate Record Analysis

```python
# 4. For each table, check for duplicate rows and report the count of duplicates.
for name, df in dataframes.items():
    display(f"\nDuplicate rows for {name}:")
    duplicate_rows = df.duplicated().sum()
    display(f"Number of duplicate rows: {duplicate_rows}")

# Duplicate order_id in orders.csv
if 'orders.csv' in dataframes:
    orders_df = dataframes['orders.csv']
    duplicate_order_ids = orders_df[orders_df.duplicated(subset='order_id', keep=False)]
    display(f"\nDuplicate order_id in orders.csv:")
    display(duplicate_order_ids)
```

# Data type Validation

```python
# 5. Validate the data types of each column in every table and convert date columns to datetime format. 
for name, df in dataframes.items():
    display(f"\nData types for {name}:")
    display(df.dtypes)

    # Convert date columns to datetime format
    for column in df.columns:
        if 'date' in column.lower():
            try:
                df[column] = pd.to_datetime(df[column], errors='coerce')
                display(f"Converted column '{column}' to datetime format.")
            except Exception as e:
                display(f"Could not convert column '{column}' to datetime format: {e}")
    display("\nUpdated data types after conversion:")
    display(df.dtypes)
```

# High Value Order Check

```python
# 6. Are there any orders above ₹5 lakh? Flag them
for name, df in dataframes.items():
    if 'final_amount' in df.columns:
        outliers = df[df['final_amount'] > 500000]
        if not outliers.empty:
            display(f"\nOutliers in {name} (final_amount > ₹5 lakh):")
            display(outliers)
        else:
            display(f"\nNo outliers found in {name} for final_amount > ₹5 lakh.")
```

# Refrential Integrity Check

```python
# 7. Verify referential integrity — do all order_ids in order_items.csv exist in orders.csv?
if 'order_items DA.csv' in dataframes and 'orders DA.csv' in dataframes:
    order_items_df = dataframes['order_items DA.csv']
    orders_df = dataframes['orders DA.csv']
    missing_order_ids = order_items_df[~order_items_df['order_id'].isin(orders_df['order_id'])]  
    if not missing_order_ids.empty:
        display(f"\nOrder IDs in order_items DA.csv that do not exist in orders DA.csv:")
        display(missing_order_ids)
    else:
        display("\nAll order_ids in order_items DA.csv exist in orders DA.csv.")
```

## Specific Checks to Run

```python
# 1. orders.csv: Check if delivered_date is always after order_date. Flag rows where it is not.
if 'orders DA.csv' in dataframes:
    df = dataframes['orders DA.csv']

    #Display actual data type 
    display(df['order_date'].dtype)
    display(df['delivered_date'].dtype)

    #change datatype to date format
    for column in df.columns:
        if 'date' in column.lower():
            df[column] = pd.to_datetime(df[column], format='%d-%m-%y', errors='coerce')
    display(df['order_date'].dtype)
    display(df['delivered_date'].dtype)

    if 'delivered_date' in df.columns:
        outliers = df[df['delivered_date'] < df['order_date']]
        if not outliers.empty:
            display("\nOutliers found in dataset:")
            display(outliers)
        else:
            display("\nNo outliers found. All delivered dates are after order dates.")


# 2. Payments.csv: Count Failed vs. Success transactions. What is the failure rate?
if 'payments DA.csv' in dataframes:
    df = dataframes['payments DA.csv']
    transaction_status = df['status'].value_counts()
    total_rows = df.shape[0]
    display(transaction_status)
    for column, count in transaction_status.items():
        if count > 0:
            percentage = (count / total_rows) * 100
            display(f"{column} rate:, {percentage:.2f}%")


 # 3. Returns.csv: Does every return have a matching order_id in orders.csv?
if 'returns DA.csv' in dataframes and 'orders DA.csv' in dataframes:
    returns_df = dataframes['returns DA.csv']
    orders_df = dataframes['orders DA.csv']
    same_order_id = returns_df[~returns_df['order_id'].isin(orders_df['order_id'])]
    if not same_order_id.empty:
        display(f"\norder_id that does not match: ")
        display(same_order_id)
    else:
        display("\nEvery return have a matching order_id.")


# 4. Inventory.csv: How many products are in 'Out of Stock' status?
if 'inventory DA.csv' in dataframes:
    df = dataframes['inventory DA.csv']
    display(df['status'].value_counts())


# 5. Customers.csv: Are there customers with total_orders = 0? 
if 'customers DA.csv' in dataframes:
    df = dataframes['customers DA.csv']
    if (df['total_orders'] == 0).any():
        display("yes, there are customers with 0 orders.")
    else:
        display("There is no one with 0 orders")
```

# Save Cleaned Datasets

```python
import os

# Create a folder to store cleaned files
output_folder = "Cleaned_Data"
os.makedirs(output_folder, exist_ok=True)

# Save every cleaned dataframe
for file_name, df in dataframes.items():

    # Remove " DA" and ".csv" from the original filename
    clean_name = file_name.replace(" DA.csv", "").replace(".csv", "")

    # Save with new name
    output_path = os.path.join(output_folder, f"cleaned_{clean_name}.csv")

    df.to_csv(output_path, index=False)

    display(f"Saved: {output_path}")

display("\nAll cleaned CSV files have been saved successfully.")
```
